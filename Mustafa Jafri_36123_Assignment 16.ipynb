```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load crop_yield.csv into crop_df
crop_df = pd.read_csv('crop_yield.csv')

# 2. Display the first 10 rows
crop_df.head(10)
```

```python
# 3. Check for missing values in each column
print("Missing Values:\n", crop_df.isnull().sum())

# 4. View data types of all columns
print("\nData Types:\n", crop_df.dtypes)
```

```python
# 5. Statistical summary of all numeric columns
print("Numeric Summary:\n", crop_df.describe())

# 6. Statistical summary of all object (categorical) columns
print("\nObject Summary:\n", crop_df.describe(include=['object', 'str']))
```

```python
# Remove rows where YIELD_KG is missing
crop_df = crop_df.dropna(subset=['YIELD_KG'])

# Fill missing values in AREA_ACRES with its mean
crop_df['AREA_ACRES'] = crop_df['AREA_ACRES'].fillna(crop_df['AREA_ACRES'].mean())

# Replace missing values in DISTRICT with 'Unknown'
crop_df['DISTRICT'] = crop_df['DISTRICT'].fillna('Unknown')

# Replace missing values in FERTILIZER_USED with 'No'
crop_df['FERTILIZER_USED'] = crop_df['FERTILIZER_USED'].fillna('No')

# Replace missing values in YIELD_VALUE with its mean
crop_df['YIELD_VALUE'] = crop_df['YIELD_VALUE'].fillna(crop_df['YIELD_VALUE'].mean())

# Convert HARVEST_DATE to datetime format
crop_df['HARVEST_DATE'] = pd.to_datetime(crop_df['HARVEST_DATE'], errors='coerce', format='%d/%m/%Y')

# Remove 'F' from FARM_ID and convert to integer
crop_df['FARM_ID'] = crop_df['FARM_ID'].str.replace('F', '').astype(int)
```
